```python
import h5py
import numpy as np

# Path to the data file
file_path = '/cis/project/organoid/May_31_2025/250314/M07359/Stimulation/000391/data.raw.h5'

# Open the HDF5 file
with h5py.File(file_path, 'r') as f:
    print("Available keys in the file:")
    print(list(f.keys()))
    
    # Explore the recordings structure
    if 'recordings' in f.keys():
        print("\nRecordings structure:")
        recordings = f['recordings']
        print(list(recordings.keys()))
        
        # Usually the first recording contains the data
        rec_key = list(recordings.keys())[0]
        print(f"\nExploring recording: {rec_key}")
        rec = recordings[rec_key]
        print(list(rec.keys()))
        
        # Look for signal data
        if 'well000' in rec.keys():
            well = rec['well000']
            print(f"\nWell structure:")
            print(list(well.keys()))
            
            if 'raw' in well.keys():
                data = well['raw'][:]
                print(f"\nData shape: {data.shape}")
                
                # Get sampling frequency
                fs = rec.attrs.get('sampling_frequency', 20000)
                print(f"Sampling frequency: {fs} Hz")
                
                # Calculate recording duration
                num_samples = data.shape[1] if data.ndim > 1 else len(data)
                duration_sec = num_samples / fs
                duration_min = duration_sec / 60
                
                print(f"Number of samples: {num_samples}")
                print(f"Recording duration: {duration_sec:.2f} seconds ({duration_min:.2f} minutes)")
```

```python
import h5py
import numpy as np

# Path to the data file
file_path = '/cis/project/organoid/May_31_2025/250314/M07359/Stimulation/000391/data.raw.h5'

# Open the HDF5 file
with h5py.File(file_path, 'r') as f:
    recordings = f['recordings']
    rec_key = list(recordings.keys())[0]
    rec = recordings[rec_key]
    
    # Check well000
    well = rec['well000']
    
    # Get start and stop times (in milliseconds)
    start_time = well['start_time'][()][0]
    stop_time = well['stop_time'][()][0]
    
    print(f"Start time: {start_time} ms")
    print(f"Stop time: {stop_time} ms")
    
    # Calculate duration in milliseconds, then convert
    duration_ms = stop_time - start_time
    duration_sec = duration_ms / 1000
    duration_min = duration_sec / 60
    
    print(f"\nRecording duration: {duration_ms} ms")
    print(f"Recording duration: {duration_sec:.2f} seconds")
    print(f"Recording duration: {duration_min:.2f} minutes")
```

```python
import h5py
import os
import glob

# Base directory
base_dir = '/cis/project/organoid/May_31_2025/250314/'

# Find all data.raw.h5 files recursively
h5_files = glob.glob(os.path.join(base_dir, '**/data.raw.h5'), recursive=True)

print(f"Found {len(h5_files)} HDF5 files\n")
print("="*80)

for file_path in sorted(h5_files):
    print(f"\nFile: {file_path}")
    print("-"*80)
    
    try:
        with h5py.File(file_path, 'r') as f:
            recordings = f['recordings']
            rec_key = list(recordings.keys())[0]
            rec = recordings[rec_key]
            
            # Check well000
            well = rec['well000']
            
            # Get start and stop times (in milliseconds)
            start_time = well['start_time'][()][0]
            stop_time = well['stop_time'][()][0]
            
            # Calculate duration
            duration_ms = stop_time - start_time
            duration_sec = duration_ms / 1000
            duration_min = duration_sec / 60
            
            print(f"Start time: {start_time} ms")
            print(f"Stop time: {stop_time} ms")
            print(f"Recording duration: {duration_ms} ms ({duration_sec:.2f} sec / {duration_min:.2f} min)")
            
    except Exception as e:
        print(f"Error processing file: {e}")
    
    print("="*80)
```
