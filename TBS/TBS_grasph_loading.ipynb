```python
import os

# Base directory to search


# Target string to search for
target = '20240116'

# Function to search for the target in the directory
def find_target(base_dir, target):
    for root, dirs, files in os.walk(base_dir):
        # Check if the target is in folder names
        for dir_name in dirs:
            if target in dir_name:
                print(f"Found folder: {os.path.join(root, dir_name)}")
        
        # Check if the target is in file names
        for file_name in files:
            if target in file_name:
                print(f"Found file: {os.path.join(root, file_name)}")
```

```python
base_dir = '/cis/project/organoid'
target = '20240116'
```

```python
find_target(base_dir, target)
```
